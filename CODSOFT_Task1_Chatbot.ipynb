```python

print("================================")
print("     IT SUPPORT CHATBOT")
print("================================")
print("Ask me about password, email, or VPN.")
print("Type 'exit' to stop the chatbot.")

while True:
    user_input = input("You: ").lower().strip()

    if user_input in ["hello", "hi", "hey"]:
        print("Bot: Hello! How can I help you?")

    elif "password" in user_input:
        print("Bot: Please use your official password reset process.")

    elif "email" in user_input or "outlook" in user_input:
        print("Bot: Check your internet connection and email access.")

    elif "vpn" in user_input:
        print("Bot: Check your VPN connection and sign-in status.")

    elif user_input == "exit":
        print("Bot: Goodbye! Have a great day.")
        break

    else:
        print("Bot: Sorry, I don't understand yet.")
        print("Bot: Please try password, email, or VPN.")
```
